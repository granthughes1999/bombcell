# Export BombCell Classification Reasons Across All Probes

This notebook uses the existing Reach15 `.env` session-loading flow to build one clean table of BombCell classification reasons for every unit across all available probes.

Optional outputs:
- `bc_classification_reason_all_probes.csv` at the selected BombCell run folder
- `cluster_bc_classificationReason.tsv` inside each `kilosort4_*` folder for Phy
- updated `cluster_info.tsv` with only `bc_classificationReason` added

## Imports

```python
from pathlib import Path
import sys
import pandas as pd

reach15_dir = Path().resolve().parent
if str(reach15_dir) not in sys.path:
    sys.path.insert(0, str(reach15_dir))

from helper_func.export_bc_classification_reasons import (
    collect_probe_dirs,
    export_all_probes,
    resolve_session_paths,
)

print('Notebook directory:', Path().resolve())
print('Reach15 root added to sys.path:', reach15_dir)
```

## Settings

```python
# Set to a specific .env path if needed. Otherwise load_dotenv() uses the default discovery behavior.
ENV_PATH = None

# 1, 2, or 3 matching the .env session blocks.
SESSION_SELECTION = 3

# Set to None to use every kilosort4_* folder found in the selected BombCell root.
# Example: ['A', 'C', 'F']
PROBES = None

# Write cluster_bc_classificationReason.tsv into each kilosort4_* folder.
WRITE_PHY_TSV = True

# Merge only bc_classificationReason into cluster_info.tsv if that file already exists.
# Keep Phy closed when using this option.
UPDATE_CLUSTER_INFO = True

OUTPUT_NAME = 'bc_classification_reason_all_probes.csv'
```

## Resolve Session and Preview Probes

```python
session_data, cfg, bombcell_root = resolve_session_paths(ENV_PATH, SESSION_SELECTION)
probe_dirs = collect_probe_dirs(bombcell_root, PROBES)

print('Recording root:', cfg['recording_root'])
print('BombCell root:', bombcell_root)
print('Session NP file:', session_data[{1: 'NP_FILE', 2: 'NP_FILE_01', 3: 'NP_FILE_02'}[SESSION_SELECTION]])
print('BombCell folder from .env:', session_data[{1: 'BOMBCELL', 2: 'BOMBCELL_01', 3: 'BOMBCELL_02'}[SESSION_SELECTION]])
print('Probes to process:', ', '.join(probe_dirs.keys()))
if UPDATE_CLUSTER_INFO:
    print('cluster_info.tsv updates assume Phy is closed.')
```

## Run Export

```python
output_path = export_all_probes(
    bombcell_root=bombcell_root,
    requested_probes=list(probe_dirs.keys()),
    write_phy_tsv=WRITE_PHY_TSV,
    update_cluster_info=UPDATE_CLUSTER_INFO,
    output_name=OUTPUT_NAME,
)

print('Combined output:', output_path)
combined_df = pd.read_csv(output_path)
display(combined_df.head(30))
print('Columns:', combined_df.columns.tolist())
print('Rows:', len(combined_df))
```

## Quick Checks

```python
summary = (
    combined_df.groupby(['probe', 'bc_unitType', 'bc_classificationReason'])
    .size()
    .rename('count')
    .reset_index()
    .sort_values(['probe', 'count'], ascending=[True, False])
)
display(summary.head(50))
```
