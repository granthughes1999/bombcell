# Run BombCell (grant_simplified)

Use this notebook to launch `run_bombcell.py` with editable parameters.

```python
from pathlib import Path
import subprocess
import sys
```

```python
# Edit these values before running (auto-detect expects this notebook inside the repository tree)
REPO_ROOT = None  # set explicitly if auto-detect fails
PY_BOMBCELL_ROOT = None
CONFIG_PATH = None
MODE = 'batch'  # 'batch' or 'single_probe'
TARGET_PROBE = 'B'  # used only when MODE == 'single_probe'
OVERWRITE = False

if REPO_ROOT is None:
    start = Path.cwd().resolve()
    for candidate in [start, *start.parents]:
        if (candidate / '.git').exists() and (candidate / 'py_bombcell').is_dir():
            REPO_ROOT = candidate
            break
    else:
        raise RuntimeError('Could not auto-detect repo root. Set REPO_ROOT manually.')

if PY_BOMBCELL_ROOT is None:
    PY_BOMBCELL_ROOT = REPO_ROOT / 'py_bombcell'

SCRIPT_PATH = PY_BOMBCELL_ROOT / 'grant_simplified/run_bombcell/run_bombcell.py'
if not SCRIPT_PATH.exists():
    raise RuntimeError(f'Expected script not found: {SCRIPT_PATH}. If folder layout changed, set paths manually.')

if CONFIG_PATH is None:
    CONFIG_PATH = PY_BOMBCELL_ROOT / 'grant_simplified/config/recording_config.json'

print('REPO_ROOT =', REPO_ROOT)
print('PY_BOMBCELL_ROOT =', PY_BOMBCELL_ROOT)
print('CONFIG_PATH =', CONFIG_PATH)
print('SCRIPT_PATH =', SCRIPT_PATH)
```

```python
cmd = [
    sys.executable,
    str(SCRIPT_PATH.relative_to(REPO_ROOT)),
    '--config',
    str(CONFIG_PATH),
    '--mode',
    MODE,
]

if MODE == 'single_probe':
    cmd.extend(['--target-probe', TARGET_PROBE])
if OVERWRITE:
    cmd.append('--overwrite')

print('Running command:')
print(' '.join(cmd))
try:
    subprocess.run(cmd, cwd=REPO_ROOT, check=True)
except subprocess.CalledProcessError as exc:
    raise RuntimeError(
        f'BombCell run failed (exit={exc.returncode}). Command: {' '.join(cmd)}\n'
        'Check CONFIG_PATH, MODE, and TARGET_PROBE values.'
    ) from exc
```
