# Run BombCell (grant_simplified)

Use this notebook to launch `run_bombcell.py` with editable parameters.

```python
from pathlib import Path
import subprocess
import sys
```

```python
# Edit these values before running
PROJECT_ROOT = None  # set explicitly if auto-detect below does not find .../py_bombcell
CONFIG_PATH = None
MODE = 'batch'  # 'batch' or 'single_probe'
TARGET_PROBE = 'B'  # used only when MODE == 'single_probe'
OVERWRITE = False

if PROJECT_ROOT is None:
    start = Path.cwd().resolve()
    for candidate in [start, *start.parents]:
        if (candidate / 'pyproject.toml').exists() and (candidate / 'grant_simplified').exists():
            PROJECT_ROOT = candidate
            break
    else:
        raise RuntimeError('Could not auto-detect py_bombcell root. Set PROJECT_ROOT manually.')

if CONFIG_PATH is None:
    CONFIG_PATH = PROJECT_ROOT / 'grant_simplified/config/recording_config.json'

print('PROJECT_ROOT =', PROJECT_ROOT)
print('CONFIG_PATH =', CONFIG_PATH)
```

```python
cmd = [
    sys.executable,
    'grant_simplified/run_bombcell/run_bombcell.py',
    '--config',
    str(CONFIG_PATH),
    '--mode',
    MODE,
]

if MODE == 'single_probe':
    cmd.extend(['--target-probe', TARGET_PROBE])
if OVERWRITE:
    cmd.append('--overwrite')

print('Running command:')
print(' '.join(cmd))
subprocess.run(cmd, cwd=PROJECT_ROOT, check=True, stdout=None, stderr=None)
```
