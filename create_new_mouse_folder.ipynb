```python
import os
import shutil
from pathlib import Path


def create_mouse_analysis_folder(
    master_dir,
    MOUSE,
    BEHAVIORAL_FOLDER,
    NP_FILE,
    NWB_FILE,
    SESSION,
    DATE,
    BOMBCELL,
    NP_FILE_01,
    NWB_FILE_01,
    DATE_01,
    SESSION_01,
    BOMBCELL_01,
    NP_FILE_02,
    NWB_FILE_02,
    DATE_02,
    SESSION_02,
    BOMBCELL_02,
    mice_output_dir="mice",
    recordings_root="H:/Grant/Neuropixels/Kilosort_Recordings",
    open_ephys_continuous_subpath="Record Node 103/experiment1/recording1/continuous",
    structure_oebin_subpath="Record Node 103/experiment1/recording1/structure.oebin",
    np20_probes="A,C,D",
    results_path="",
):
    """
    Copy Master -> mice/MOUSE, create .env, and sync run files used by BC_open_ephys_unified.ipynb.
    """
    master_dir = Path(master_dir).resolve()
    mice_output_dir = Path(mice_output_dir).resolve()
    mouse_folder_path = mice_output_dir / MOUSE

    if mouse_folder_path.exists():
        print(f"Mouse folder already exists: {mouse_folder_path} (skip to avoid overwrite).")
        return

    mice_output_dir.mkdir(parents=True, exist_ok=True)
    shutil.copytree(master_dir, mouse_folder_path)

    # Keep new folders aligned with the latest run notebook + prep helper in repo root.
    repo_root = master_dir.parent
    sync_pairs = [
        (repo_root / "BC_open_ephys_unified.ipynb", mouse_folder_path / "run_bc" / "BC_open_ephys_unified.ipynb"),
        (repo_root / "nwb_data_prep.py", mouse_folder_path / "run_bc" / "nwb_data_prep.py"),
    ]
    for src, dst in sync_pairs:
        if src.exists():
            dst.parent.mkdir(parents=True, exist_ok=True)
            shutil.copy2(src, dst)
            print(f"Synced: {dst}")

    env_file_path = mouse_folder_path / ".env"
    env_lines = [
        f"MOUSE={MOUSE}",
        f"BEHAVIORAL_FOLDER={BEHAVIORAL_FOLDER}",
        f"NP_FILE={NP_FILE}",
        f'NWB_FILE="{NWB_FILE}"',
        f"DATE={DATE}",
        f"SESSION={SESSION}",
        f"BOMBCELL={BOMBCELL}",
        f"NP_FILE_01={NP_FILE_01}",
        f'NWB_FILE_01="{NWB_FILE_01}"',
        f"DATE_01={DATE_01}",
        f"SESSION_01={SESSION_01}",
        f"BOMBCELL_01={BOMBCELL_01}",
        f"NP_FILE_02={NP_FILE_02}",
        f'NWB_FILE_02="{NWB_FILE_02}"',
        f"DATE_02={DATE_02}",
        f"SESSION_02={SESSION_02}",
        f"BOMBCELL_02={BOMBCELL_02}",
        "",
        "# Optional defaults used to auto-generate per-session grant configs",
        f"RECORDINGS_ROOT={recordings_root}",
        f"OPEN_EPHYS_CONTINUOUS_SUBPATH={open_ephys_continuous_subpath}",
        f"STRUCTURE_OEBIN_SUBPATH={structure_oebin_subpath}",
        f"NP20_PROBES={np20_probes}",
        f"RESULTS_PATH={results_path}",
    ]
    env_file_path.write_text("\n".join(env_lines) + "\n", encoding="utf-8")

    print(f"Mouse folder created: {mouse_folder_path}")
```
